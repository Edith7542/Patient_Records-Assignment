```python
# Import the pandas library
import pandas as pd

# Store patients as a list of dictionaries
patients = [
    {"Patient ID": "201", "Name": "Peter", "Age": 30, "Diagnosis": "Headache"},
    {"Patient ID": "202", "Name": "Mary", "Age": 25, "Diagnosis": "Check-up"},
    {"Patient ID": "203", "Name": "Amina", "Age": 18, "Diagnosis": "Cough"},
    {"Patient ID": "204", "Name": "John", "Age": 40, "Diagnosis": "UTI"},
    {"Patient ID": "205", "Name": "Grace", "Age": 50, "Diagnosis": "URTI"}
]

# Convert the patient records into a DataFrame
df = pd.DataFrame(patients)

# Display the DataFrame as a table
print(df)
```

```python
# Create a new patient dictionary
new_patient = {
    "Patient ID": input("Enter patient ID: "),
    "Name": input("Enter patient name: "),
    "Age": int(input("Enter patient age: ")),
    "Diagnosis": input("Enter patient diagnosis: ")
}

# Add the patient to the list
patients.append(new_patient)

# Rebuild the DataFrame to include the new patient
df = pd.DataFrame(patients)

# Display the updated table and count
print(df)
print("Total patients:", len(df))
```
